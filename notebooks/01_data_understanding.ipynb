```python
from nltk.tokenize import sent_tokenize

df["num_sentences"] = df["message"].apply(
    lambda x: len(sent_tokenize(x))
)
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="num_characters",
    hue="label",
    bins=60,
    kde=True,
    element="step"
)

plt.title("Character Count Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="num_words",
    hue="label",
    bins=50,
    kde=True,
    element="step"
)

plt.title("Word Count Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="num_sentences",
    hue="label",
    bins=20,
    kde=True,
    element="step"
)

plt.title("Sentence Count Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="label",
    y="num_characters"
)

plt.title("Character Count by Label")
plt.show()
```
